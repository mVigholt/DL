```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer
import torch, re

# Load the GPT -2 model and tokenizer
tokenizer = GPT2Tokenizer.from_pretrained('gpt2')
model = GPT2LMHeadModel.from_pretrained('gpt2')
model.eval()

# Establish news labels
news_cats = {
    0 : "World",
    1 : "Sports",
    2 : "Business",
    3 : "Sci/Tech"
}

# Prepare the few-shot prompt with examples
few_shot_prompt = (
    "Categorize the following articles into one of the four categories: 0, 1, 2, 3:\n"
    "Article: 'Mortars Mark Opening of Iraqi Political Conference BAGHDAD (Reuters) - Insurgents fired mortars at a meeting where Iraqi leaders met to pick an interim national assembly Sunday, killing at least two people in a grim reminder of the country's tortured path toward democracy.'\n"
    "Category: '0'\n\n"
    "Article: 'Galaxy, Crew Play to 0-0 Tie (AP) AP - Kevin Hartman made seven saves for Los Angeles, and Jon Busch had two saves for Columbus as the Galaxy and Crew played to a 0-0 tie Saturday night.'\n"
    "Category: '1'\n\n"
    "Article: 'Apple to open second Japanese retail store this month (MacCentral) MacCentral - Apple Computer Inc. will open its second Japanese retail store later this month in the western Japanese city of Osaka, it said Thursday.'\n"
    "Category: '2'\n\n"
    "Article: 'Oil and Economy Cloud Stocks' Outlook NEW YORK (Reuters) - Soaring crude prices plus worries about the economy and the outlook for earnings are expected to hang over the stock market next week during the depth of the summer doldrums.'\n"
    "Category: '3'\n\n"
)

def get_gpt2_prediction(article):
    prompt = few_shot_prompt + f"Article: {article}\nCategory:"
    inputs = tokenizer.encode(prompt, return_tensors='pt')
    
    outputs = model.generate(
        inputs,
        max_length = inputs.shape[1] + 10,
        do_sample = False,
        num_beams = 5,
        early_stopping = True,
        output_attentions = True
    )

    generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
    print(f"Generated text:\n{generated_text}")

    # Extract the translation from the generated text
    # Set up regex
    pattern = r"'([^']+)'(?:$|\n)"
    matches = re.findall(pattern, generated_text)
    # if len(matches) > 0:
    #     result = matches[-1]
    # else:
    #     print("No matches in text.\n")

    # predicted_translation = result

    # return predicted_translation
```
