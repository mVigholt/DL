## Object masks from prompts with SAM

The Segment Anything Model (SAM) predicts object masks given prompts that indicate the desired object. The model first converts the image into an image embedding that allows high quality masks to be efficiently produced from a prompt.

The `SamPredictor` class provides an easy interface to the model for prompting the model. It allows the user to first set an image using the `set_image` method, which calculates the necessary image embeddings. Then, prompts can be provided via the `predict` method to efficiently predict masks from those prompts. The model can take as input both point and box prompts, as well as masks from the previous iteration of prediction.

```python
import numpy as np
import torch
import matplotlib.pyplot as plt
import cv2
```

```python
def show_mask(mask, ax, random_color=False):
    if random_color:
        color = np.concatenate([np.random.random(3), np.array([0.6])], axis=0)
    else:
        color = np.array([30/255, 144/255, 255/255, 0.6])
    h, w = mask.shape[-2:]
    mask_image = mask.reshape(h, w, 1) * color.reshape(1, 1, -1)
    ax.imshow(mask_image)

def show_points(coords, labels, ax, marker_size=375):
    pos_points = coords[labels==1]
    neg_points = coords[labels==0]
    ax.scatter(pos_points[:, 0], pos_points[:, 1], color='green', marker='*', s=marker_size, edgecolor='white', linewidth=1.25)
    ax.scatter(neg_points[:, 0], neg_points[:, 1], color='red', marker='*', s=marker_size, edgecolor='white', linewidth=1.25)

def show_box(box, ax):
    x0, y0 = box[0], box[1]
    w, h = box[2] - box[0], box[3] - box[1]
    ax.add_patch(plt.Rectangle((x0, y0), w, h, edgecolor='green', facecolor=(0,0,0,0), lw=2))
```

```python
def plot_image_with_segmentation(image, segmentation, ax=None):
    """
    Plots an image with overlayed segmentation mask

    Returns: plt.fig and ax objects
    """
    if ax is None:
        fig = plt.figure(figsize=(8, 8))
        ax = fig.add_subplot(111)
        ax.axis("off")

    ax.imshow(image.squeeze(), cmap="gray")
    mask = np.ma.masked_where(segmentation == 0, segmentation)
    ax.imshow(mask.squeeze(), cmap="Set1", alpha=0.5)
    return plt.gcf(), ax


def load_npz_dataset(path, keys=('x_train', 'y_train', 'x_val', 'y_val', 'x_test', 'y_test')):
    archive = np.load(path)
    return [archive.get(key) for key in keys]
```

```python
def map_interval(image, from_min, from_max, to_min, to_max):
    """
    Map values from [from_min, from_max] to [to_min, to_max]
    """
    from_range = from_max - from_min
    to_range = to_max - to_min
    # scaled = np.array((image - from_min) / float(from_range), dtype=float)
    scaled = (image - from_min) / float(from_range)
    return to_min + (scaled * to_range)
```

```python
# Load train/val/test data
x_train, y_train, x_val, y_val, x_test, y_test = load_npz_dataset(data_fn)

# TODO:
# Bring images into the correct format for SAM:
# Image shape: (N, H, W, C=3)
# Mask shape: (N, H, W)
# Values: [0, 255] (uint8)

# YOUR CODE HERE

def preprocess_data(images, masks):

    # Duplicate grayscale channel to convert into RGB format
    # from (N, H, W, 1) to (N, H, W, 3)
    if images.shape[-1] == 1:
        images = np.repeat(images, 3, axis=-1)

    # Scale images to range [0, 255]
    imgMin, imgMax = np.min(images), np.max(images)
    images = ( (images-imgMin) / (imgMax-imgMin) * 255).astype(np.uint8)
    images= np.clip(images, 0, 255)
    
    # Scale masks to range [0, 255]
    masks = (masks * 255).astype(np.uint8)

    # Remove last mask dimension
    if masks.shape[-1] == 1:
        masks = masks.squeeze(axis=-1)

    return images, masks


x_train_prep, y_train_prep = preprocess_data(x_train, y_train)
x_val_prep, y_val_prep = preprocess_data(x_val, y_val)
x_test_prep, y_test_prep = preprocess_data(x_test, y_test)

#¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤#
# Check the processed data
# print(f"x_train_processed shape: {x_train_prep.shape}")
# print(f"y_train_processed shape: {y_train_prep.shape}")
# print(f"x_train_processed dtype: {x_train_prep.dtype}")
# print(f"y_train_processed dtype: {y_train_prep.dtype}")
# print(f"np.min(x_train_processed): {np.min(x_train_prep)}")
# print(f"np.max(x_train_processed): {np.max(x_train_prep)}")
#¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤¤#

# your data should pass the following asserts
assert x_train_prep.shape == (112, 256, 256, 3)
assert y_train_prep.shape == (112, 256, 256)
assert x_val_prep.shape == (12, 256, 256, 3)
assert y_val_prep.shape == (12, 256, 256)
assert x_test_prep.shape == (123, 256, 256, 3)
assert y_test_prep.shape == (123, 256, 256)

assert x_train_prep.dtype == y_train_prep.dtype == np.uint8
assert np.min(x_train_prep) == 0
assert np.max(x_train_prep) == 255

# Plot an example
fig, ax = plot_image_with_segmentation(x_train_prep[0], y_train_prep[0])
plt.show()
```

# Evaluation loop

```python
import torchmetrics
f1 = torchmetrics.F1Score(task="binary")

f1_scores = []

for img, mask_gt in zip(x_val_prep, y_val_prep):
    
    # Set image for prediction
    predictor.set_image(img)

    # The prediction process itself
    masks, scores, logits = predictor.predict(
    point_coords=input_points,
    point_labels=input_label,
    multimask_output=True,
    )

    # Select best mask based on model's own scores
    best_mask_idx = np.argmax(scores)
    predicted_mask = masks[best_mask_idx]

    # Convert masks to binary [0, 1]
    predicted_mask_binary = (predicted_mask > 0).astype(np.uint8)
    mask_gt_binary = (mask_gt > 0).astype(np.uint8)

    # Convert binary masks to tensors with the same dtype
    predicted_mask_tensor = torch.tensor(predicted_mask_binary, dtype=torch.int32)
    mask_gt_tensor = torch.tensor(mask_gt_binary, dtype=torch.int32)

    # Calculate F1 score
    f1_score = f1(predicted_mask_tensor, mask_gt_tensor)
    f1_scores.append(f1_score.item())

    
```

```python
f1_scores_numpy = np.array(f1_scores)
mean_f1 = np.mean(f1_scores_numpy)
std_f1 = np.std(f1_scores_numpy)

print(f"Mean F1 score: {mean_f1:.4f}")
print(f"Standard deviation: {std_f1:.4f}")
```
