### Generator for space of strings

```python
import random

class ABCGenerator:
    def __init__(self, sample_num, max_length, abc_percentage):
        self.sample_num = sample_num
        self.max_length = max_length
        self.abc_percentage = abc_percentage

    def generate_string(self, label):
        length = random.randint(1, self.max_length)
        if label == 1:
            a_count = random.randint(1, max(1, length // 3))
            b_count = a_count
            c_count = a_count
        else:
            a_count = random.randint(1, length)
            b_count = random.randint(1, length - a_count)
            c_count = length - (a_count + b_count)
        
        abc_string = 'a' * a_count + 'b' * b_count + 'c' * c_count
        abc_string = ''.join(random.sample(abc_string, len(abc_string)))
        
        return f"<{abc_string},{label}>"

    def generate_samples(self):
        num_label_1 = int(self.sample_num * self.abc_percentage)
        num_label_0 = self.sample_num - num_label_1
        
        samples = [self.generate_string(1) for _ in range(num_label_1)]
        samples += [self.generate_string(0) for _ in range(num_label_0)]
        
        random.shuffle(samples)
        return samples

# Example usage:
generator = ABCGenerator(1000, 20, 0.5)
samples = generator.generate_samples()
for sample in samples:
    print(sample)
```
